```python
import matplotlib.pyplot as plt
x=[]
y1=[]
y2=[]
x=range(-100,100,10)
for i in x:
    y1.append(i**2)
for i in x:
    y2.append(-i**2)
plt.plot(x,y1,color='green')
plt.plot(x,y2,color='maroon')
plt.xlabel('x')
plt.ylabel('y')
plt.title('simple graph')
plt.axhline(color='red')
plt.axhline(color='green')
plt.savefig("linegraph.jpg")
plt.grid(True)
plt.show()
```
